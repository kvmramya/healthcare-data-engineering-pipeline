# Bronze Layer Ingestion — Healthcare Data Engineering

## Project Overview

This notebook demonstrates the **Bronze layer** of a simple healthcare data engineering pipeline using **PySpark and Delta Lake**.

The Bronze layer preserves incoming data in a structured format with minimal transformation so that it can be used as the foundation for downstream **Silver** transformation and **Gold** analytics layers.

> **Data Privacy Note:** All patient and visit records in this notebook are **synthetic sample data created for demonstration purposes**. They do not represent real patients or real clinical records.

## Objectives

- Create synthetic patient and hospital-visit datasets using PySpark.
- Inspect the resulting DataFrame schemas.
- Perform a basic record-count validation.
- Persist the datasets as Delta tables.
- Validate the Bronze tables using SQL.

## Technology Stack

- Python
- Apache Spark / PySpark
- Delta Lake
- SQL
- Databricks

## 1. Create Synthetic Patient Data

The first dataset contains basic demographic attributes such as patient ID, name, age, gender, and city.

The data is intentionally synthetic and is included only to demonstrate the ingestion workflow.


````python
data = [
    ("P001", "Ravi", 45, "M", "Hyderabad"),
    ("P002", "Anita", 32, "F", "Delhi"),
    ("P003", "John", 61, "M", "Mumbai"),
    ("P004", "Priya", 28, "F", "Bengaluru"),
    ("P005", "Arjun", 54, "M", "Chennai"),
    ("P006", "Sneha", 39, "F", "Hyderabad"),
    ("P007", "Rahul", 67, "M", "Pune"),
    ("P008", "Meena", 43, "F", "Delhi"),
    ("P009", "Kiran", 50, "M", "Mumbai"),
    ("P010", "Divya", 35, "F", "Hyderabad")
]

columns = ["patient_id", "name", "age", "gender", "city"]

patients_df = spark.createDataFrame(data, columns)

patients_df.show(truncate=False)
````

## 2. Inspect Patient Schema

Schema inspection helps confirm the columns and Spark-inferred data types before the data is persisted.


````python
patients_df.printSchema()
````

## 3. Basic Data Quality Check

A simple record count is performed as an initial ingestion validation.


````python
print("Number of patients:", patients_df.count())
````

## 4. Write Patient Data to the Bronze Layer

The patient DataFrame is stored as a **Delta table** named `bronze_patients`.

`overwrite` mode is used here because this notebook is a demonstration pipeline and recreates the sample dataset each time it is run.


````python
patients_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze_patients")
````

## 5. Validate the Bronze Patient Table

Read the newly created Delta table using Spark SQL to confirm that the Bronze table is available and contains the expected records.


````python
%sql
SELECT *
FROM bronze_patients;
````
