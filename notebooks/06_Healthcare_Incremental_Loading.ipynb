```python
from pyspark.sql import Row
from pyspark.sql.functions import current_timestamp
```

```python
audit_data = [
    Row(
        pipeline_name="Healthcare Incremental Patient Load",
        target_table="workspace.healthcare.silver_patients",
        status="SUCCESS"
    )
]
```

```python
audit_df = spark.createDataFrame(
    audit_data
).withColumn(
    "execution_timestamp",
    current_timestamp()
)

display(audit_df)
```

```python
audit_df.write \
    .format("delta") \
    .mode("append") \
    .saveAsTable(
        "workspace.healthcare.pipeline_audit_log"
    )
```

```python
audit_log = spark.table(
    "workspace.healthcare.pipeline_audit_log"
)

display(audit_log)
```
